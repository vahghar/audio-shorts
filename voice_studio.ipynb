```python
import os
from pydub import AudioSegment, effects

os.makedirs("temp_clips", exist_ok=True)

# Optional but recommended: paste the exact words spoken in your 2 reference clips
REF_TEXTS = {
    "rogan.wav": "",   # e.g., "It's wild how fast this technology is moving..."
    "kamath.wav": "",  # e.g., "I think most people underestimate how long cycles take..."
}

podcast = AudioSegment.empty()

for i, item in enumerate(episode_data):
    speaker = item["speaker_name"]
    text = item["text"]
    is_joe = "joe" in item["speaker"].lower()

    ref_audio = item.get("voice_file", "rogan.wav" if is_joe else "kamath.wav")
    speed = item.get("speed", 1.0)
    interrupt = item.get("interrupt", False)
    pause_ms = item.get("pause_after_ms", 300)

    print(f"[{i+1}/{len(episode_data)}] Voicing {speaker} ({speed}x){' [CUT-IN]' if interrupt else ''}...")
    clip_path = f"temp_clips/clip_{i}.wav"

    f5tts.infer(
        ref_file=ref_audio,
        ref_text=REF_TEXTS.get(ref_audio, ""),
        gen_text=text,
        speed=speed,
        file_wave=clip_path
    )

    clip = AudioSegment.from_wav(clip_path)
    clip = effects.normalize(clip)
    clip = clip.pan(-0.15 if is_joe else 0.15)

    # Check if the NEXT turn is an interruption so we don't add dead air before a cut-in
    next_is_interrupt = (i + 1 < len(episode_data)) and episode_data[i + 1].get("interrupt", False)

    if i == 0:
        podcast = clip
    elif interrupt and len(podcast) > 350 and len(clip) > 350:
        # Overlap the start of this clip over the end of the previous speaker
        podcast = podcast.append(clip, crossfade=300)
    else:
        podcast += clip

    if not next_is_interrupt and i < len(episode_data) - 1:
        podcast += AudioSegment.silent(duration=pause_ms)

podcast.export("final_podcast.mp3", format="mp3", bitrate="192k")
!ffmpeg -y -i final_podcast.mp3 -af "highpass=f=80,equalizer=f=120:t=q:w=1:g=2.5,acompressor=threshold=-18dB:ratio=3:attack=5:release=50,loudnorm" studio_podcast.mp3
print("Done!")
```
